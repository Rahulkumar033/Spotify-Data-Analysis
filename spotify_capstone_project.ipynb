```python
# Task 6 : Examine correlations between different features (e.g., energy vs. danceability).
import plotly.express as px
sample_df = df_tracks.sample(2000, random_state=42)
fig = px.scatter(
    sample_df,
    x="danceability",
    y="energy",
    title="Energy vs Danceability (Sampled Data)",
    opacity=0.5)

fig.update_traces(marker_color="orange", marker_size=5)
```

```python
# Task 7 : Analyze user preferences and listening habits.
import plotly.express as px
top_genres = (
    df_genres.groupby("genres")["popularity"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index())
```

```python
# Task 7 : Analyze user preferences and listening habits.
import plotly.express as px
fig = px.bar(
    top_genres,
    x="genres",
    y="popularity",
    title="User Preferred Genres on Spotify",
    labels={"genres": "Genre", "popularity": "Average Popularity"})
fig.update_traces(marker_color="pink")
```

```python
# Task 7 : Analyze user preferences and listening habits.
import plotly.express as px
px.scatter(
    df_tracks,
    x="danceability",
    y="energy",
    size="popularity",
    color="valence",
    title="User Preferences and Listening Habits (Single View)",
    labels={
        "danceability": "Danceability",
        "energy": "Energy",
        "popularity": "Popularity",
        "valence": "Mood (Valence)"},
    opacity=0.4)
```

```python
# Task 4.2: Use advanced visualization techniques (e.g., heatmaps, pair plots) to uncover deeper insights.
import plotly.express as px
features = [
    "danceability",
    "energy",
    "tempo",
    "valence",
    "loudness",
    "acousticness",
    "speechiness"]

corr_matrix = df_tracks[features].corr()
```

```python
# Task 9 : Use advanced visualization techniques (e.g., heatmaps, pair plots) to uncover deeper insights.
import plotly.express as px
px.imshow(
    corr_matrix,
    text_auto=True,
    color_continuous_scale="RdBu",
    title="Correlation Heatmap of Spotify Audio Features")
```

```python
# Task 4.2: Use advanced visualization techniques (e.g., heatmaps, pair plots) to uncover deeper insights.
import plotly.express as px
pair_features = [
    "danceability",
    "energy",
    "valence",
    "acousticness",
    "popularity"]
```

```python
# Task 9 : Use advanced visualization techniques (e.g., heatmaps, pair plots) to uncover deeper insights.
import plotly.express as px
px.scatter_matrix(
    df_tracks,
    dimensions=pair_features,
    title="Pair Plot of Spotify Audio Features",
    opacity=0.3,color_discrete_sequence=["purple"])
```
